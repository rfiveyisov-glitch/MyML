```python
import  numpy  as np 
import matplotlib.pyplot as plt
from sklearn.datasets  import  make_blobs
from  sklearn.cluster  import KMeans
from  sklearn.metrics import  silhouette_samples,silhouette_score
```

```python
n_samples = 300
n_features  = 5
n_clusters = 4
random_state  = 42

X,y = make_blobs(n_samples=n_samples,
                 centers=n_clusters,
                 n_features=n_features,
                 random_state=random_state)


kmeans = KMeans(n_clusters=n_clusters,
                random_state=random_state)

kmeans.fit(X)
```

```python
labels  = kmeans.labels_
labels
```

```python
centroids = kmeans.cluster_centers_
centroids
```

```python
within_cluster_variance  = kmeans.inertia_
within_cluster_variance
```

```python
plt.figure(figsize=(10,5))
plt.scatter(X[:,0],
            X[:,1],
            c =labels,
            cmap = 'viridis',
            marker='o',
            edgecolors='k',
            alpha =0.2)

for  i , center in  enumerate(centroids):
    circle = plt.Circle(
                        center,1.5, 
                        color='black',
                        linestyle='--',
                        fill=False 
                        )
    plt.gca().add_artist(circle)

    plt.text(
             center[0], center[1],f'Cluster{i+1} ',
             fontsize = 15 , 
             color = 'purple',
             ha = 'right',
            )
plt.scatter(centroids[:,0],
            centroids[:,1],
            c='red',
            marker='x',
            s=200,
            # labels= 'centroids',
            alpha =0.5)

plt.title('Cluster Identified by K_means')
plt.xlabel("Feature 1 ")
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

```python
def plot_elbow_method(X, max_k=10):
    inertias = []
    K_range = range(1, max_k+1)
    
    for k in K_range:
        kmeans = KMeans(n_clusters=k, random_state=42)
        kmeans.fit(X)
        inertias.append(kmeans.inertia_)  # WCSS dəyəri
    
    # Qrafikin çəkilməsi
    plt.figure(figsize=(8,5))
    plt.plot(K_range, inertias, 'bo-')
    plt.xlabel('Klasterlərin sayı (K)')
    plt.ylabel('Inertiya (WCSS)')
    plt.title('Optimal K-nın təyini üçün Elbow Metodu')
    plt.grid(True)
    plt.show()

# Funksiyanın istifadəsi
plot_elbow_method(X, max_k=10)
```

```python


def plot_silhouette_scores(data,
                            max_k=10):
    silhouette_scores = []
    K_range = range(2, max_k+1)  # Silhouette ən azı 2 klaster üçün hesablanır
    
    for k in K_range:
        kmeans = KMeans(n_clusters=k,
                         random_state=42)
        cluster_labels = kmeans.fit_predict(data)
        score = silhouette_score(data, cluster_labels)
        silhouette_scores.append(score)
    
    plt.figure(figsize=(8,5))
    plt.plot(K_range, silhouette_scores, 'ro-')
    plt.xlabel('Klasterlərin sayı (K)')
    plt.ylabel('Silhouette Skoru')
    plt.title('Optimal K-nın təyini üçün Silhouette Metodu')
    plt.grid(True)
    plt.axhline(y=np.mean(silhouette_scores),
                color='gray',
                  linestyle='--',
                    alpha=0.7)
    plt.show()
    
    best_k = K_range[np.argmax(silhouette_scores)]
    print(f'Ən yüksək Silhouette skoru: {max(silhouette_scores):.3f} (K = {best_k})')
    return silhouette_scores

```

```python
# Funksiyanın istifadəsi
scores = plot_silhouette_scores(X, max_k=10)
```

```python
def silhouette_plot(X,
                    labels,
                    n_clusters,
                    title="Silhouette Plot for Clustering",
                    fig_size=(8,6),
                    x_lab='Coefficient Values',
                    y_lab='Cluster',
                    silh_colors=None,
                    x_axis_line_col='black'):
    
    # 1. Hər nöqtə üçün siluet skorunu hesabla
    sil_vals = silhouette_samples(X, labels)
    
    # 2. Ümumi orta siluet skorunu hesabla
    mean_score = silhouette_score(X, labels)
    
    # 3. Qrafik üçün konteyner yarat
    _, ax = plt.subplots(figsize=fig_size)
    
    # 4. Klasterləri çəkmək üçün başlanğıc mövqe
    y_lower = 10
    
    # 5. Rəngləri təyin et (standart: tab10 rəng xəritəsi)
    if silh_colors is None:
        silh_colors = plt.cm.tab10.colors
    
    # 6. Hər klaster üçün vizuallaşdırma
    for i in range(n_clusters):
        # a) Cari klasterə aid olan nöqtələrin siluet skorlarını seç
        cluster_sil_vals = sil_vals[labels == i]
        
        # b) Skorları çəkmək üçün sırala
        cluster_sil_vals.sort()
        
        # c) Y oxunda klasterin başlanğıc və bitiş nöqtələrini təyin et
        y_upper = y_lower + len(cluster_sil_vals)
        
        # d) Klaster üçün rəng seç (dairəvi indeksləmə)
        colour = silh_colors[i % len(silh_colors)]
        
        # e) Klasterin siluet qrafikini çək (doldurulmuş sahə)
        ax.fill_betweenx(
            np.arange(y_lower, y_upper),  # Y oxu dəyərləri
            0,                            # X oxunun başlanğıc nöqtəsi
            cluster_sil_vals,             # X oxunun bitiş nöqtələri
            alpha=0.7,                    # Şəffaflıq
            color=colour                  # Rəng
        )
        
        # f) Klaster nömrəsini qrafikə əlavə et
        ax.text(-0.05,                      # X mövqeyi
                y_lower + 0.5 * len(cluster_sil_vals),  # Y mövqeyi (ortada)
                str(i)                      # Klaster nömrəsi
        )
        
        # g) Növbəti klaster üçün başlanğıc nöqtəni təyin et
        y_lower = y_upper + 10  # Klasterlər arasında 10 vahid boşluq
    
    # 7. Ümumi orta siluet skorunu şaquli xətt ilə göstər
    ax.axvline(x=mean_score,
               color=x_axis_line_col,
               linestyle="-")
    
    # 8. Qrafikin digər parametrlərini təyin et
    ax.set_title(title)
    ax.set_xlabel(x_lab)
    ax.set_ylabel(y_lab)
    ax.set_yticks([])  # Y oxundakı rəqəmləri gizlət
    
    # 9. Şəbəkə xətlərini əlavə et
    grid_lines = np.arange(-0.1, 1.1, 0.1)
    
    # 10. Qrafiki göstər
    plt.show()
```

```python
silhouette_plot( X,
                 labels,
                   n_clusters,
                silh_colors=['red', 'black', 'navy','orange'])
```

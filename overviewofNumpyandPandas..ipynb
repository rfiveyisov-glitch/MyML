```python
heights = np.array([150, 160, 170, 180, 190])
weights = np.array([50, 60, 65, 80, 90])

print("Heights:", heights)
print("Weights:", weights)
```

```python
heights_in_meters = heights / 100
print("Heights in meters:", heights_in_meters)
```

```python
scaled_weights = weights * 2
print("Scaled Weights (multiplied by 2):", scaled_weights)
```

```python
combined = heights * weights
print("Combined:", combined)
```

```python
mean_height = np.mean(heights)
mean_weight = np.mean(weights)
print("Mean Height:", mean_height)
print("Mean Weight:", mean_weight)
```

```python
std_height = np.std(heights)
std_weight = np.std(weights)
print("Standard Deviation of Heights:", std_height)
print("Standard Deviation of Weights:", std_weight)
```

```python
var_height = np.var(heights)
var_weight = np.var(weights)
print("Variance of Heights:", var_height)
print("Variance of Weights:", var_weight)
```

```python
median_height = np.median(heights)
median_weight = np.median(weights)
print("Median Height:", median_height)
print("Median Weight:", median_weight)
```

```python
min_height = np.min(heights)
max_height = np.max(heights)
min_weight = np.min(weights)
max_weight = np.max(weights)
print("Min Height:", min_height, "Max Height:", max_height)
print("Min Weight:", min_weight, "Max Weight:", max_weight)
```

```python
import numpy as np
houses = np.array([
                    [3, 2000, 10],
                    [2, 1500, 20],
                    [4, 2500, 5],
                    ])
```

```python
updated_bedrooms = houses[:, 0] + 1  #  1-ci  sutuna + 1 
print("Updated bedrooms after broadcasting:\n", updated_bedrooms)
```

```python
updated_age = houses[:, 2] - 5  #  3-cu sutundan - 5
print("Updated house ages after broadcasting:\n", updated_age)
```

```python
adjusted_size = houses[:, 1] * 1.1
print("Adjusted house sizes (sq ft) by 10% increase:\n", adjusted_size)
```

--------------------------------------------------------------

```python
print(houses.shape)
```

```python
reshaped_houses = houses.reshape(-1)   #  (9,0)
print("Reshaped houses (flattened):\n", reshaped_houses)
```

```python
reshaped_houses_1x9 = houses.reshape(1, 9)  # (1,9)
print("Reshaped houses into 1x9:\n", reshaped_houses_1x9.shape)
```

---------------------------------------

```python

more_houses = np.array([[5,
                         3000,
                         15]])
```

```python
joined_houses = np.concatenate((houses,more_houses),
                               axis=0) # alt alta birlesdirir
print("Joined houses (by rows):\n", joined_houses)
```

```python
extra_data = np.array([[100000],
                       [120000],
                       [140000]])

joined_columns = np.concatenate((houses, extra_data),
                                 axis=1)  # yan-yana birlesdirir
print("Joined houses with extra column (price):\n", joined_columns)
```

```python
stacked_houses_vertically = np.vstack((houses, more_houses))    #  alt-alta birlesdirir
print("Vertically stacked houses:\n", stacked_houses_vertically)
```

```python
stacked_houses_horizontally = np.hstack((houses, extra_data))   #  yan-yana  birledirir
print("Horizontally stacked houses with extra column:\n", stacked_houses_horizontally)
```

```python
weights = np.array([50, 100,-20])
bias = 30000
predicted_prices = np.dot(houses, weights) + bias
print("Predicted prices:", predicted_prices)
```

```python
house_norms = np.linalg.norm(houses, axis=1, keepdims=True)  # ||x|| = sqrt( bedrooms^2 +  size^2 + age^2)
normalized_houses = houses / house_norms   # x / ||x||

predicted_prices_normalized = np.dot(normalized_houses, weights) + bias
print("Predicted prices (normalized):", predicted_prices_normalized)
```

## Singular value decomposition

```python
U, S, Vt = np.linalg.svd(houses, full_matrices=False)

print("U matrix:\n", U,'\n')
print("Singular values:\n", S,'\n')
print("V transpose matrix:\n", Vt,'\n')
```

Covariance


```python
cov_matrix = np.cov(houses, rowvar=False)
print("Covariance matrix:\n", cov_matrix)
```

Correlation matrix


```python
corr_matrix = np.corrcoef(houses, rowvar=False)
print("Pearson correlation matrix:\n", corr_matrix)
```
